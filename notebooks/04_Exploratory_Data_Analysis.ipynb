# 04 - Exploratory Data Analysis (EDA)

## Objective

Explore the cleaned marketing dataset to understand trends, seasonality,
marketing spend behaviour, relationships, and potential issues before feature
engineering and model development.

> In industry, EDA is used to generate business hypotheses, not just charts.


```

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path

plt.style.use("default")

ROOT = Path.cwd()
DATA = ROOT/"data"/"processed"/"marketing_mix_cleaned.csv"

df = pd.read_csv(DATA, parse_dates=["Week"])

media_cols = [
    "Google_Search","Google_Display","Meta","Instagram",
    "YouTube","TV","Radio","Influencer","Affiliate","Email"
]

df.head()
```

## Dataset Overview

```
print(df.shape)
display(df.head())
```

## Data Types

```
display(df.dtypes.to_frame('dtype'))
```

## Summary Statistics

```
display(df.describe().T)
```

## Sales Trend

```
plt.figure(figsize=(14,4))
plt.plot(df["Week"],df["Sales"])
plt.title("Weekly Sales Trend")
plt.grid(True)
plt.show()
```

## Monthly Sales

```
monthly=df.set_index("Week")["Sales"].resample("M").sum()
monthly.plot(figsize=(14,4),title="Monthly Sales")
plt.show()
```

## Marketing Spend

```
df[media_cols].sum().sort_values().plot.barh(figsize=(8,5))
plt.title("Total Marketing Spend by Channel")
plt.show()
```

## Correlation Heatmap

```
corr=df.select_dtypes("number").corr()
plt.figure(figsize=(10,8))
plt.imshow(corr,aspect="auto")
plt.xticks(range(len(corr)),corr.columns,rotation=90)
plt.yticks(range(len(corr)),corr.columns)
plt.colorbar()
plt.title("Correlation Matrix")
plt.show()
```

## Sales vs Google Search

```
plt.scatter(df["Google_Search"],df["Sales"],alpha=.6)
plt.xlabel("Google Search Spend")
plt.ylabel("Sales")
plt.show()
```

## Holiday Impact

```
display(df.groupby("Holiday")["Sales"].agg(["mean","median","max"]))
```

## Channel Spend Share

```
share=df[media_cols].sum()
share=(share/share.sum()*100).round(2)
display(share.sort_values(ascending=False).to_frame("Spend %"))
```

## Weekly Spend

```
df["Total_Media_Spend"]=df[media_cols].sum(axis=1)
plt.figure(figsize=(14,4))
plt.plot(df["Week"],df["Total_Media_Spend"])
plt.title("Weekly Total Media Spend")
plt.show()
```

## Price Distribution

```
df["Price"].hist(bins=30)
plt.title("Price Distribution")
plt.show()
```

## Competitor Spend

```
plt.figure(figsize=(14,4))
plt.plot(df["Week"],df["Competitor_Spend"])
plt.title("Competitor Spend")
plt.show()
```

## Sales Distribution

```
df["Sales"].hist(bins=30)
plt.title("Sales Distribution")
plt.show()
```
