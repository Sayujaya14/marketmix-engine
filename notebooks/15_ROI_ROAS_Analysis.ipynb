# 15 - ROI, ROAS & Marketing Efficiency Analysis

## Objective

Evaluate the financial effectiveness of each marketing channel.

This notebook demonstrates how to calculate:

- Return on Investment (ROI)
- Return on Ad Spend (ROAS)
- Marketing Efficiency Ratio (MER)
- Cost per Incremental Sale (CPIS)
- Marginal ROAS (illustrative)
- Executive KPI summary

> In production, these calculations are typically based on channel contributions from the final MMM.


```

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path

ROOT=Path.cwd()
DATA=ROOT/"data"/"processed"/"marketing_mix_model_ready.csv"

df=pd.read_csv(DATA)

media=[
"Google_Search_Hill","Meta_Hill","TV_Hill"
]

# Simple synthetic revenue attribution for demonstration
revenue={}
for m in media:
    revenue[m]=df[m].sum()*np.random.uniform(2.5,5.5)

metrics=[]
for m in media:
    spend=df[m].sum()
    rev=revenue[m]
    roi=(rev-spend)/(spend+1e-9)
    roas=rev/(spend+1e-9)
    cpis=spend/max(rev/1000,1)
    mer=rev/(df[media].sum().sum()+1e-9)
    metrics.append([m,spend,rev,roi,roas,cpis,mer])

kpi=pd.DataFrame(metrics,columns=[
    "Channel","Spend","Attributed_Revenue",
    "ROI","ROAS","CPIS","MER"
])

display(kpi)
```

## 1. ROI Comparison

```

plt.figure(figsize=(8,4))
plt.bar(kpi["Channel"],kpi["ROI"])
plt.title("ROI by Channel")
plt.ylabel("ROI")
plt.xticks(rotation=20)
plt.show()
```

## 2. ROAS Comparison

```

plt.figure(figsize=(8,4))
plt.bar(kpi["Channel"],kpi["ROAS"])
plt.title("ROAS by Channel")
plt.ylabel("ROAS")
plt.xticks(rotation=20)
plt.show()
```

## 3. Cost per Incremental Sale

```

plt.figure(figsize=(8,4))
plt.bar(kpi["Channel"],kpi["CPIS"])
plt.title("Cost per Incremental Sale")
plt.ylabel("CPIS")
plt.xticks(rotation=20)
plt.show()
```

## 4. Marginal ROAS (Illustrative)

```

kpi["Marginal_ROAS"]=kpi["ROAS"]*np.random.uniform(0.85,1.05,len(kpi))
display(kpi[["Channel","ROAS","Marginal_ROAS"]])
```

## 5. Executive KPI Dashboard

```

dashboard=pd.DataFrame({
    "KPI":[
        "Total Marketing Spend",
        "Attributed Revenue",
        "Average ROI",
        "Average ROAS",
        "Best Channel"
    ],
    "Value":[
        f"₹{kpi['Spend'].sum():,.0f}",
        f"₹{kpi['Attributed_Revenue'].sum():,.0f}",
        round(kpi["ROI"].mean(),2),
        round(kpi["ROAS"].mean(),2),
        kpi.sort_values("ROAS",ascending=False).iloc[0]["Channel"]
    ]
})
display(dashboard)
```

# Business Interpretation

### ROI
Measures overall profitability.

\[
ROI = \frac{Revenue - Spend}{Spend}
\]

### ROAS
Measures advertising efficiency.

\[
ROAS = \frac{Revenue}{Advertising\ Spend}
\]

### MER
Marketing Efficiency Ratio compares total attributed revenue to total marketing spend.

### CPIS
Cost required to generate one incremental sale.

## Executive Recommendations

- Increase investment in consistently high-ROAS channels.
- Review channels with high spend but weak ROI.
- Validate attribution before making budget decisions.
- Combine these metrics with channel contribution analysis.
